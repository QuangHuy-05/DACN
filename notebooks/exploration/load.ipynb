```python
# Cài đặt (nếu chạy trên Colab chưa có): 
# !pip install vietnamadminunits postal

from vietnamadminunits import parse_address as vn_parse
from vietnamadminunits import convert_address as vn_convert
from postal.parser import parse_address as lp_parse

# 1. Chuẩn bị 6 nhóm Test Cases x 2 địa chỉ
test_categories = {
    "1. ĐỊA CHỈ ĐẦY ĐỦ (HỆ MỚI)": [
        "70 Nguyễn Sỹ Sách, Phường Tân Sơn, Thành phố Hồ Chí Minh",
        "Ngã 4 Xóm Ao Dài, Xã Thanh Trì, Thủ đô Hà Nội"
    ],
    "2. ĐỊA CHỈ CŨ BẨN (VIẾT TẮT, THIẾU DẤU)": [
        "đường 15, long bình, q.9, hcm",
        "59 nguyen sy sach, p15, tan binh, tp. hcm"
    ],
    "3. ĐỊA CHỈ THIẾU TRƯỜNG": [
        "169/8A Thoại Ngọc Hầu, Quận Tân Phú", # Thiếu Phường, Tỉnh/TP
        "Ngõ 133 Thái Hà, Hà Nội"               # Thiếu Phường, Quận
    ],
    "4. ĐỊA CHỈ MÔ TẢ BẰNG MỐC": [
        "Gần ngã tư Hàng Xanh hướng Bình Thạnh, TP HCM",
        "Đối diện chợ Bà Chiểu, đường Bạch Đằng, Bình Thạnh"
    ],
    "5. ĐỊA CHỈ LAI (TRỘN LẪN CŨ - MỚI)": [
        "Phường An Khánh, Quận 2, TPHCM",                   # Phường mới, Quận cũ
        "Khu công nghệ cao, Thành phố Thủ Đức, Quận 9"      # Tên TP mới, Quận cũ
    ],
    "6. ĐỊA CHỈ CẶP ÁNH XẠ 2 CHIỀU (SÁP NHẬP)": [
        "02 Lê Đại Hành, Phường 15, Quận 11, TP. Hồ Chí Minh", # Hệ cũ (sẽ sáp nhập thành P. Phú Thọ)
        "Phường Phú Thọ, Quận 11, Thành phố Hồ Chí Minh"       # Hệ mới (cần tra ngược về P.13 hoặc P.15 cũ)
    ]
}

def test_models():
    print("=" * 80)
    print("ĐÁNH GIÁ ĐỐI CHỨNG: LIBPOSTAL vs VIETNAMADMINUNITS")
    print("=" * 80)

    for category, addresses in test_categories.items():
        print(f"\n\n{category}")
        print("-" * 80)
        
        for addr in addresses:
            print(f"\n📍 [INPUT]: {addr}")
            
            # --- TEST LIBPOSTAL (Đa quốc gia) ---
            print("   [LIBPOSTAL]:")
            try:
                lp_res = lp_parse(addr)
                # Format kết quả libpostal
                lp_formatted = ", ".join([f"{label}: '{val}'" for val, label in lp_res])
                print(f"      -> {lp_formatted}")
            except Exception as e:
                print(f"      -> ❌ Lỗi: {e}")

            # --- TEST VIETNAMADMINUNITS (Dựa luật) ---
            print("   [VN_ADMIN_UNITS]:")
            try:
                # Ưu tiên parse theo hệ mới 2025
                vn_res = vn_parse(addr, mode='FROM_2025', keep_street=True)
                print(f"      -> ✅ Parse (FROM_2025): {vn_res.get_address()}")
            except Exception as e1:
                try:
                    # Nếu thất bại, thử convert từ hệ cũ sang hệ mới
                    vn_res_conv = vn_convert(addr, mode='CONVERT_2025')
                    print(f"      -> 🔄 Convert (LEGACY -> 2025): {vn_res_conv.get_address()}")
                except Exception as e2:
                    print(f"      -> ❌ BÓ TAY (Lỗi: Không khớp được cấu trúc)")

test_models()
```

```python
import osmium
from collections import Counter
import json

class OSMExplorer(osmium.SimpleHandler):
    def __init__(self, sample_size=10):
        super(OSMExplorer, self).__init__()
        self.samples = []
        self.sample_size = sample_size
        self.field_counter = Counter()

    def process_tags(self, tags):
        # Chuyển danh sách tags của OSM thành Dictionary (Key-Value)
        tag_dict = {k: v for k, v in tags}
        
        # LỌC: Chỉ quan tâm đến những đối tượng có ít nhất 1 tag liên quan đến địa chỉ (addr:)
        if any(k.startswith('addr:') for k in tag_dict):
            
            # Thống kê xem trường nào được dùng nhiều nhất
            for k in tag_dict:
                if k.startswith('addr:'):
                    self.field_counter[k] += 1
            
            # Thu thập đủ 10 mẫu đầu tiên để in ra màn hình
            if len(self.samples) < self.sample_size:
                self.samples.append(tag_dict)

    def node(self, n):
        self.process_tags(n.tags)

    def way(self, w):
        self.process_tags(w.tags)

# Khởi tạo công cụ khám phá
explorer = OSMExplorer(sample_size=10)
file_path = 'vietnam-250101.osm.pbf' # Đổi thành tên file 2025 của bạn

print(f"🔍 Đang quét và phân tích cấu trúc file: {file_path}...")
try:
    # Quét file
    explorer.apply_file(file_path)
    
    # 1. IN RA THỐNG KÊ CÁC TRƯỜNG PHỔ BIẾN NHẤT
    print("\n" + "="*80)
    print("📊 BẢNG XẾP HẠNG CÁC TRƯỜNG ĐỊA CHỈ TRONG FILE")
    print("="*80)
    for field, count in explorer.field_counter.most_common(15):
        print(f" - {field:<20}: xuất hiện {count:>7} lần")

    # 2. IN RA TOP 10 MẪU RAW (NGUYÊN BẢN)
    print("\n" + "="*80)
    print("📦 TOP 10 MẪU RAW DATA ĐẦU TIÊN (ĐỂ BẠN XEM CẤU TRÚC)")
    print("="*80)
    for i, sample in enumerate(explorer.samples, 1):
        # In đẹp (Pretty print) theo format JSON
        print(f"\n--- Mẫu {i} ---")
        print(json.dumps(sample, indent=2, ensure_ascii=False))

except FileNotFoundError:
    print(f"❌ Không tìm thấy file {file_path}. Vui lòng kiểm tra lại đường dẫn!")
except Exception as e:
    print(f"❌ Có lỗi xảy ra: {e}")
```
